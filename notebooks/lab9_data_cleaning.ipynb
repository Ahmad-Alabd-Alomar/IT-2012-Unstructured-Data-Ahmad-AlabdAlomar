```python
import os
import sys
sys.path.append(os.path.abspath(os.path.join('..')))

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

from src.cleaning.missing_handler import handle_missing_values, report_missing
from src.cleaning.string_cleaner import clean_strings
from src.cleaning.deduplicator import remove_duplicates
from src.cleaning.type_converter import convert_types
from src.analytics.regex_ops import validate_and_clean_regex
from src.cleaning.validator import validate_data

# Load the raw data exported from Lab 8
df = pd.read_csv("../data/raw/analytics_export.csv")
print(f"Original Data Shape: {df.shape}")
df.head(2)
```

```python
print("--- 1. Deduplication ---")
df_deduped = remove_duplicates(df)
print(f"Rows remaining: {len(df_deduped)}")
```

```python
print("--- 2. Missing Values ---")
print("Before cleaning:")
display(df_deduped.isnull().sum()[df_deduped.isnull().sum() > 0])

df_filled = handle_missing_values(df_deduped)

print("\nAfter cleaning:")
display(df_filled.isnull().sum()[df_filled.isnull().sum() > 0])
```

```python
print("--- 3. String & Regex Cleaning ---")
df_strings = clean_strings(df_filled)
df_regex = validate_and_clean_regex(df_strings)

# Show a sample of cleaned titles (assuming 'title' exists)
if 'title' in df_regex.columns:
    display(df_regex[['title']].head(5))
```

```python
print("--- 4. Type Conversion ---")
df_final = convert_types(df_regex)
display(df_final.dtypes)

print("\n--- 5. Final Validation ---")
is_valid = validate_data(df_final)

if is_valid:
    output_path = "../data/processed/cleaned/cleaned_data.csv"
    import os
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    df_final.to_csv(output_path, index=False)
    print(f"\n✅ Clean dataset saved to {output_path}!")
```
